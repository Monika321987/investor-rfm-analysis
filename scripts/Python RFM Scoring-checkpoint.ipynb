```python
import sys
import pandas as pd
from sqlalchemy import create_engine
```

```python
DB_USER = "postgres"
DB_PASSWORD = "Bball1987"
DB_HOST = "localhost"
DB_PORT = "5432"
DB_NAME = "resume"

DATABASE_URL = f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(DATABASE_URL)

sql_query = """
WITH max_date_cte AS (
	SELECT MAX(transaction_timestamp) AS global_max_date
	FROM financial_transactions
),
last_balance AS (
	SELECT DISTINCT ON (account_id) account_id, account_balance
	FROM financial_transactions
	ORDER BY account_id, transaction_timestamp DESC
)
SELECT f.account_id, 
	/* Calculate days of inactivity relative to lastest date in dataset.*/
	(SELECT global_max_date::date FROM max_date_cte) - MAX(f.transaction_timestamp)::date AS recency,
	COUNT(f.transaction_id) AS frequency,
	l.account_balance AS total_aum
FROM financial_transactions f
JOIN last_balance l 
	ON f.account_id = l.account_id
	GROUP BY f.account_id, l.account_balance
	ORDER BY f.account_id;
"""
print("Data from PostgreSQL...")
df = pd.read_sql(sql_query, con=engine)
df.head()
```

```python
print("Scoring RFM metrics...")

# Recency: Ranked 5 to 1, LOWER numbers being most recent are better. Lowest gets scored 5.
df['R_Score'] = pd.qcut(df['recency'].rank(method='first', ascending=True), q=5, labels=[5,4,3,2,1])

# Frequency & Monetary: Ranked 5 to 1 with HIGHER amounts better. Highest gets scored 5.
df['F_Score'] = pd.qcut(df['frequency'].rank(method='first'), q=5, labels=[1,2,3,4,5])
df['M_Score'] = pd.qcut(df['total_aum'].rank(method='first'), q=5, labels=[1,2,3,4,5])

#combine scores into a string. Show first 5 rows of scores
df['RFM_String'] = df['R_Score'].astype(str) + df['F_Score'].astype(str) + df['M_Score'].astype(str)
df[['account_id','R_Score','F_Score','M_Score','RFM_String']].head()

#Business Segment Mapping
def assign_segment(row):
    r = int(row['R_Score'])
    f = int(row['F_Score'])
    m = int(row['M_Score'])

    if r >= 4 and f >= 4 and m >= 4:
         return 'VIP Whale Investor'
    elif r >= 3 and f >= 3 and m <= 2:
        return 'Active / Low Capital User'
    elif r <= 2 and f >= 3 and m >= 3:
        return 'High Value Churn Risk'
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Dormant Account'
    elif r >= 4 and f <= 2:
        return 'New Account Profile'
    else:
        return 'Mid-Tier Mid-Market'

print("Investors mapped to business segments...")

df['Investor_Segment'] = df.apply(assign_segment, axis=1)
df.head()
```

```python
#export dataframe to csv
df.to_csv('financials_segmented.cvs', index=False)
```
